```python
# removing null values
# customer id

df.dropna(subset=['Customer_ID'],inplace=True)
df.isnull().sum()
```

```python
df['Age'].unique()
```

```python
import re
def extract_age(age):
  age_num=re.findall('[0-9]+',str(age))

  if len(age_num)>0:
    return int(age_num[0])
  else:
    return age

df['Age'] = df['Age'].apply(lambda x :extract_age(x))
```

```python
df['Age'].unique()
```

```python
df_age =df[df['Age'] != 'nan years']['Age']

age_median = int(df_age.dropna().astype('int64').median())
age_median
```

```python
df_age.dropna()
```

```python
df.replace('nan years',age_median,inplace=True)
```

```python
df['Age'].fillna(age_median,inplace=True)
```

```python
df['Age'].unique()
```

```python
df.isnull().sum()
```

```python
df['Purchase_Amount'].fillna(df['Purchase_Amount'].median(),inplace=True)
df['Purchase_Amount'].unique()
```

```python
df.isnull().sum()
```

```python
df['Feedback_Score'].fillna(df['Feedback_Score'].mode()[0],inplace=True)
```

```python
df.isnull().sum()
```

```python
for col in ['Gender','City','Country']:
  df[col].fillna(df[col].mode()[0],inplace=True)
```

```python
df['Last_Purchase_Date'].ffill(inplace=True)
```

```python
# df['Gender'].unique()
df['Gender'] = df['Gender'].str.lower().str.strip()
df['Gender'].replace({'m': 'male','f': 'female'},inplace=True)
```

```python
# df['Country'].unique()
df['Country'] = df['Country'].str.lower().str.strip()
df['Country'].replace({'ind': 'india'},inplace=True)
```

```python
# df['City'].unique()
df['City'] = df['City'].str.lower().str.strip()
df['Country'].replace({'ind': 'india'},inplace=True)
```

```python
df.info()
```

```python
df.drop_duplicates(inplace=True)
df['Customer_ID'].value_counts()
```

```python
df.drop_duplicates(subset=['Customer_ID'], keep= 'first',inplace=True)
```

```python
df['Customer_ID'].value_counts()
```

```python
# correcct data type
df['Age'].astype('int64')

df['Signup_Date']=pd.to_datetime(df['Signup_Date'])

df['Last_Purchase_Date']=pd.to_datetime(df['Last_Purchase_Date'])
```

```python
# handling outliers
cols =['Age','Purchase_Amount','Feedback_Score']

plt.figure(figsize=(6,3))

for i,col in enumerate(cols,1):
  plt.subplot(1,3,i)
  sns.boxplot(y = df[col],color = 'blue',width =0.3)
  plt.title(f'Boxplot of {col}',fontsize = 12)
  plt.ylabel('')
  plt.grid(axis='y',linestyle ='--',alpha = 0.6)

plt.tight_layout()
plt.show()
```

```python
cols =['Age','Purchase_Amount','Feedback_Score']

plt.figure(figsize=(6,3))

for i,col in enumerate(cols,1):
  plt.subplot(1,3,i)
  sns.boxplot(y = df_clean[col],color = 'green',width =0.3)
  plt.title(f'Boxplot of {col}',fontsize = 12)
  plt.ylabel('')
  plt.grid(axis='y',linestyle ='--',alpha = 0.6)

plt.tight_layout()
plt.show()
```

```python
# wht is the age distribution of customer

plt.figure(figsize=(8,4))

sns.histplot(
    data=df_clean,
    x='Age',
    bins=15,
    kde=True,
    color='steelblue'
)

plt.title('Customer Age Distribution')
plt.xlabel('Age')
plt.ylabel('Number of Customers')
plt.grid(axis='y',alpha =0.3)

plt.show()
```

```python
# which city generates thw highest purchase revenue

city_sales = (
    df_clean.groupby('City')['Purchase_Amount']
    .agg(['sum','mean','count'])
    .sort_values('sum',ascending=False)
)

print(city_sales)
```

```python
plt.figure(figsize=(6,3))

sns.barplot(
    data=city_sales.reset_index(),
    x='sum',
    y='City',
    color='steelblue'
)

plt.title('Total purchase ampunt by city')
plt.xlabel('total purchase amount')
plt.ylabel('City')

plt.show()
```

```python
# is there any relation btw purchase amount and feedback score

feedback_analysis =(
    df_clean.groupby('Feedback_Score')['Purchase_Amount']
    .agg(['mean','sum','count'])
)

print(feedback_analysis)
```

```python
# who are the highest value customer
top_customers = df_clean.sort_values("Purchase_Amount", ascending=False).head(10)[
    [ "Name", "City"]
]

print(top_customers)
```

```python
# total purchase amount by gender

gender_purchase =(
    df_clean.groupby('Gender')['Purchase_Amount']
    .sum()
    .sort_values(ascending=False)
)

print(gender_purchase)
# bar graph
```

```python
# 2. How is Purchase_Amount distributed — is it skewed, and are there visible outliers?

print(df_clean["Purchase_Amount"].describe())
plt.hist(df["Purchase_Amount"], bins=40, color="darkorange", edgecolor="black")
plt.title("Distribution of Purchase Amount")
plt.xlabel("Purchase Amount")
plt.ylabel("Frequency")
plt.show()
```

```python
# Do male and female customers differ in average spending?

avg_by_gender = df_clean.groupby("Gender")["Purchase_Amount"].mean()
print(avg_by_gender)
```

```python
# Is there any relationship between a customer's Age and how much they spend?

correlation = df["Age"].corr(df["Purchase_Amount"])
print("Correlation:", correlation)

plt.scatter(df["Age"], df["Purchase_Amount"], alpha=0.3, s=10)
plt.title("Age vs Purchase Amount")
plt.xlabel("Age")
plt.ylabel("Purchase Amount")
plt.show()
```

```python
# . How have customer signups trended over time — growing, flat, or declining?

df_clean["Signup_Date"] = pd.to_datetime(df_clean["Signup_Date"])
signups_by_month = df_clean.set_index("Signup_Date").resample("ME").size()
print(signups_by_month)

signups_by_month.plot(kind="line", marker="o")
plt.title("Customer Signups Over Time (Monthly)")
plt.xlabel("Month")
plt.ylabel("Number of Signups")
plt.show()
```

```python
# 6. Is there a correlation between Feedback_Score and Purchase_Amount?

correlation = df_clean["Feedback_Score"].corr(df_clean["Purchase_Amount"])
print("Correlation:", correlation)
```

```python
# What does the age distribution of your customer base look like?

plt.hist(df_clean["Age"], bins=30, color="teal", edgecolor="black")
plt.title("Age Distribution of Customers")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

```python
# How does the gender split vary across cities?

gender_city = df_clean.groupby(["City", "Gender"]).size().unstack(fill_value=0)
print(gender_city)

gender_city.plot(kind="bar", stacked=True)
plt.title("Gender Split by City")
plt.xlabel("City")
plt.ylabel("Number of Customers")
plt.show()
```

```python
# On average, how many days pass between signup and last purchase?

df_clean["Last_Purchase_Date"] = pd.to_datetime(df["Last_Purchase_Date"])
df_clean["Days_To_Last_Purchase"] = (df_clean["Last_Purchase_Date"] - df_clean["Signup_Date"]).dt.days
print(df_clean["Days_To_Last_Purchase"].mean())
```

```python
# What share of total revenue comes from the top 10% of spenders?

sorted_amounts = df_clean["Purchase_Amount"].sort_values(ascending=False)
top_10_count = int(len(sorted_amounts) * 0.10)
top_10_revenue = sorted_amounts.head(top_10_count).sum()
total_revenue = sorted_amounts.sum()
print(f"Top 10% contribute {top_10_revenue / total_revenue * 100:.1f}% of total revenue")
```
